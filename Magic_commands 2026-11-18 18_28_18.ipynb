####How to run a Spark SQL Queries inside a notebook using %sql magic command

```python
%sql create table if not exists default.cities3(id int,city string);
insert into default.cities3 values(3,'Mumbai'),(4,'Lucknow');
select count(*) from cities3;
select * from cities3;
```

```python
%sql select * from cities3
```

```python
%sql
update cities3 set city='Kolkata' where id=4;
select * from cities3;
```

```python
%sql
--We are performing DDL operations which inturn access the metadata from Unitycatalog..
show create table cities3;
```
